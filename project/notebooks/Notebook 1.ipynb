<h1>Kinematische analyse van het paraplu-mechanisme</h1>

<p>
In deze notebook wordt de volledige <b>kinematische analyse</b> uitgevoerd van een vlak paraplu-mechanisme.
Het mechanisme wordt gemodelleerd als een systeem van starre links verbonden via ideale gewrichten.
</p>

<p>
De analyse omvat:
</p>

<ul>
<li>de topologische modellering van het mechanisme;</li>
<li>de bepaling van de mobiliteit;</li>
<li>het opstellen van sluitingsvergelijkingen;</li>
<li>de numerieke positie-analyse;</li>
<li>de snelheids- en versnellingsanalyse;</li>
<li>de interpretatie van het mechanisch gedrag over het bewegingsbereik.</li>
</ul>

<p>
De gebruikte aanpak is gebaseerd op de methode van de <span style="color: Maroon;"><b>sluitingsvergelijkingen</b></span>,
zoals behandeld in <b>les 2</b>.  
Deze methode laat toe om de configuratie van complexe mechanismen systematisch en numeriek te bepalen.
</p>

<p>
De nadruk in deze notebook ligt niet alleen op het uitvoeren van berekeningen, maar ook op het
<b>begrijpen en interpreteren</b> van het gedrag van het mechanisme.
</p>

</div>

<h3>Situering van het mechanisme</h3>

<p>
Het beschouwde systeem is een <b>vlak mechanisme</b>, waarbij alle bewegingen zich in één vlak afspelen.
Het paraplu-mechanisme wordt gemodelleerd als een verzameling van <b>starre lichamen (links)</b>
die onderling verbonden zijn door <b>ideale gewrichten</b>.
</p>

<p>
De structuur van het mechanisme wordt volledig bepaald op basis van de <b>gegeven joint table</b>.
Deze tabel beschrijft welke links met elkaar verbonden zijn en via welk type gewricht.
Er wordt <b>geen extra informatie</b> uit de schets afgeleid die niet expliciet in de joint table aanwezig is.
</p>

<p>
<span style="color: Maroon;"><b>Belangrijk</b></span>: kruisingen van staven in de figuur impliceren

<p>

<h3>Topologische modellering: links en gewrichten</h3>

<p>
Een mechanisme wordt in de kinematica eerst beschreven aan de hand van zijn <b>topologie</b>,
namelijk de manier waarop de verschillende <b>links</b> met elkaar verbonden zijn door
<b>gewrichten</b>.
Deze topologische beschrijving gebeurt <b>vóór</b> het opstellen van kinematische of dynamische vergelijkingen.
</p>

<p>
Een <span style="color: Maroon;"><b>link</b></span> is een star lichaam waarvan de vorm tijdens de beweging niet verandert.
In deze analyse worden alle blauwe elementen van het mechanisme als <b>starre lichamen</b> beschouwd.
Wanneer een element in de figuur labels draagt met hetzelfde nummer maar met suffixen zoals
<b>3a</b> en <b>3b</b>, of <b>8a</b> en <b>8b</b>, dan worden deze beschouwd als
delen van <b>éénzelfde star lichaam</b>, dus niet als aparte links.
Deze deelstukken stellen vaste geometrische deelvectoren voor binnen dezelfde link.
</p>

<p>
Een <span style="color: Maroon;"><b>gewricht</b></span> verbindt twee links met elkaar en legt beperkingen op aan hun relatieve beweging.
In dit mechanisme komen uitsluitend <b>revolute joints</b> (draaigewrichten) en
één <b>prismatic joint</b> (schuifgewricht) voor.
Alle gewrichten worden als <b>ideaal</b> beschouwd:
er is dus geen speling, geen flexibiliteit en geen wrijving in de topologische beschrijving.
</p>

<p>
De topologie van het mechanisme wordt in deze notebook <b>niet</b> afgeleid uit de visuele figuur alleen,
maar expliciet vastgelegd via de opgegeven joint table.
Indien stangen elkaar kruisen in de figuur, betekent dit <b>niet</b> automatisch dat er daar een gewricht aanwezig is.
Enkel de hieronder opgegeven verbindingen worden mechanisch als gewricht beschouwd.
</p>

<h3>Mobiliteit van het mechanisme</h3>

<p>
De mobiliteit (of het aantal vrijheidsgraden) van een vlak mechanisme wordt bepaald met de formule van
<span style="color: Maroon;"><b>Grübler/Kutzbach</b></span>.
</p>

</div>

$$
M = 3(n - 1) - 2f_1 - f_2
$$

<div style="width: 76ch;">

<p>
waarbij:
</p>

<ul>
<li><b>n</b>: aantal links, inclusief de grond</li>
<li><b>f<sub>1</sub></b>: aantal één-vrijheidsgraad gewrichten (revolute en prismatic)</li>
<li><b>f<sub>2</sub></b>: aantal twee-vrijheidsgraden gewrichten (hier niet aanwezig)</li>
</ul>

<h4>Toepassing op het mechanisme</h4>

<p>
Uit de topologische modellering volgt:
</p>

<ul>
<li>aantal links: n = 8</li>
<li>alle gewrichten zijn één-vrijheidsgraad gewrichten, dus f<sub>2</sub> = 0</li>
<li>aantal gewrichten volgens de joint table: f<sub>1</sub> = 10</li>
</ul>

<p>
Invullen in de formule geeft:
</p>

</div>

$$
M = 3(8 - 1) - 2 \cdot 10 = 21 - 20 = 1
$$

<div style="width: 76ch;">

<h4>Besluit</h4>

<p>
Het mechanisme heeft <span style="color: Maroon;"><b>één vrijheidsgraad</b></span>.
</p>

<h4>Fysische interpretatie</h4>

<p>
Dit betekent dat de volledige configuratie van het mechanisme bepaald wordt door
<b>één onafhankelijke invoer</b>.
In dit geval is dat de translatie van de schuiver langs de centrale stok.
</p>

<ul>
<li>één actuator volstaat om het volledige mechanisme aan te drijven;</li>
<li>de schuiverpositie <em>s</em> bepaalt alle hoeken en posities in het mechanisme;</li>
<li>alle andere configuratievariabelen zijn functies van deze ene invoercoördinaat.</li>
</ul>

<p>
Wanneer de schuiver beweegt, worden alle andere links via de gewrichten mee gedwongen
in een unieke configuratie. Dat bevestigt dat de keuze van de schuiver als primaire invoercoördinaat
zowel fysisch als kinematisch correct is.
</p>

</div>

<h3>Keuze van grond, invoer en output</h3>

<p>
Voor de kinematische analyse moeten expliciet keuzes gemaakt worden voor:
</p>

<ul>
<li>de <b>grondlink</b>;</li>
<li>de <b>invoercoördinaat</b>;</li>
<li>de <b>taakuitgang</b>.</li>
</ul>

<p>
Deze keuzes bepalen hoe het mechanisme wiskundig beschreven wordt en moeten daarom
duidelijk gemotiveerd worden.
</p>

<h4>Keuze van de grond</h4>

<p>
Als grond wordt <b>link 1</b> gekozen, dit is de centrale vaste stok van de paraplu.
</p>

<p>
Deze keuze is natuurlijk omdat:
</p>

<ul>
<li>link 1 niet beweegt in de werkelijkheid;</li>
<li>alle andere links ten opzichte van deze stok bewegen;</li>
<li>dit een stabiel en fysisch logisch referentiekader oplevert.</li>
</ul>

<p>
Het vaste coördinatenstelsel wordt dus gekoppeld aan link 1.
</p>

<h4>Keuze van de invoercoördinaat</h4>

<p>
De invoer van het mechanisme wordt gekozen als de translatie van de schuiver (link 2)
langs de centrale stok.
Deze wordt voorgesteld door de variabele <em>s</em>.
</p>

<p>
Deze keuze is fysisch en kinematisch logisch omdat:
</p>

<ul>
<li>de schuiver de beweging van het mechanisme initieert (openen/sluiten van de paraplu);</li>
<li>het mechanisme één vrijheidsgraad heeft, dus één invoer volstaat;</li>
<li>alle andere posities en hoeken bepaald worden door deze ene variabele;</li>
<li>deze invoer rechtstreeks overeenkomt met de werkelijke bediening van een paraplu.</li>
</ul>

<p>
In de praktijk kan deze translatie gerealiseerd worden door een motor in de centrale stok
die via een spindel de schuiver op en neer beweegt.
Dit verandert niets aan de topologie van het mechanisme en wordt daarom niet expliciet gemodelleerd.
</p>

<h4>Keuze van de output</h4>

<p>
Als primaire output wordt de positie van punt <b>K</b> gekozen, genoteerd als:

</div>

$$

$$

<div style="width: 76ch;">

<p>
Punt K ligt aan het uiteinde van de buitenste rib (link 8) en beschrijft rechtstreeks
de geometrische opening van de paraplu.
</p>

<p>
Deze keuze is geschikt omdat:
</p>

<ul>
<li>de beweging van K een direct beeld geeft van het openen en sluiten van de paraplu;</li>
<li>deze output fysisch intuïtief interpreteerbaar is;</li>
<li>de volledige vorm van het mechanisme zichtbaar wordt via de positie van K;</li>
<li>deze grootheid rechtstreeks bruikbaar is voor grafieken en visualisaties.</li>
</ul>

<p>
Als secundaire output kan eventueel de hoek van link 8 beschouwd worden,
maar deze wordt niet als primaire grootheid gebruikt omdat ze minder direct
de globale geometrie van de paraplu beschrijft.
</p>

</ul>

<p>
Deze keuzes zijn fysisch logisch, consistent met de werking van het mechanisme

<h3>Keuze van assenstelsel en configuratievariabelen</h3>

<p>
Om de kinematische analyse op te stellen, moet eerst een geschikt
<b>assenstelsel</b> gekozen worden en moeten de <b>configuratievariabelen</b>
duidelijk gedefinieerd worden.
Deze keuzes bepalen hoe de sluitingsvergelijkingen worden opgesteld en hebben
een grote invloed op de eenvoud van de verdere analyse.
</p>

<h4>Keuze van het assenstelsel</h4>

<p>
We kiezen een vast cartesisch assenstelsel gekoppeld aan de grond (link 1).
De oorsprong wordt geplaatst in punt <b>C</b>, het vaste gewricht bovenaan de centrale stok.
</p>

<p>
De assen worden als volgt gekozen:
</p>

<ul>
<li>de <b>x-as</b> is horizontaal;</li>
<li>de <b>y-as</b> is verticaal;</li>
<li>de positieve y-richting wijst naar boven langs de centrale stok.</li>
</ul>

<p>
De schuiver (link 2) beweegt langs deze verticale as.
Zijn positie kan dus rechtstreeks beschreven worden door één coördinaat langs de y-as.
</p>

<h4>Motivatie van deze keuze</h4>

<p>
Deze keuze van assenstelsel vereenvoudigt de analyse om verschillende redenen:
</p>

<ul>
<li>de beweging van de schuiver gebeurt volledig langs de y-as, waardoor geen projectie nodig is;</li>
<li>de invoercoördinaat kan rechtstreeks als een cartesiaanse coördinaat geïnterpreteerd worden;</li>
<li>de vaste link (link 1) valt samen met de y-as, wat de geometrie vereenvoudigt;</li>
<li>de sluitingsvergelijkingen krijgen een eenvoudige en fysisch interpreteerbare vorm;</li>
<li>de beweging (openen/sluiten) blijft intuïtief interpreteerbaar als een op/neer beweging.</li>
</ul>

<h4>Configuratievariabelen</h4>

<p>
De configuratie van het mechanisme wordt beschreven door:
</p>

<ul>

<li>de oriëntatiehoeken van de bewegende links.</li>

<p>
Voor elke starre link wordt één globale oriëntatiehoek gedefinieerd:
</p>

<ul>

De onbekende configuratievariabelen zijn:
</p>

<ul>

<h3>Onafhankelijke kinematische lussen van het mechanisme</h3>

<p>
Om de kinematica van het mechanisme te beschrijven, wordt gebruik gemaakt van
<span style="color: Maroon;"><b>sluitingsvergelijkingen</b></span>.
Deze worden opgesteld voor een verzameling van <b>onafhankelijke kinematische lussen</b>.
</p>

<p>
Een lus is een gesloten keten van links en gewrichtspunten waarbij men,
vertrekkend van een bepaald punt, via opeenvolgende verbonden links
terugkeert naar hetzelfde punt.
Voor elke dergelijke gesloten keten geldt dat de som van de verplaatsingsvectoren nul is.
</p>

<h4>Aantal onafhankelijke lussen</h4>

<p>
Voor een verbonden vlak mechanisme met enkel lagere paren is het aantal
onafhankelijke lussen gelijk aan:
</p>

<p style="text-align:center;">
<b>L = j - n + 1</b>
</p>

<p>
waarin:
</p>

<ul>
<li><b>j</b>: aantal gewrichten,</li>
<li><b>n</b>: aantal links, inclusief de grond.</li>
</ul>

<p>
Voor dit mechanisme geldt:
</p>

<ul>
<li><b>n = 8</b>,</li>
<li><b>j = 10</b>.</li>
</ul>

<p>
Dus:
</p>

<p style="text-align:center;">
<b>L = 10 - 8 + 1 = 3</b>
</p>

<p>
Er zijn dus <span style="color: Maroon;"><b>drie onafhankelijke kinematische lussen</b></span> nodig
om alle geometrische beperkingen van het mechanisme volledig te beschrijven.
</p>

<h4>Keuze van de drie onafhankelijke lussen</h4>

<p>
Een consistente keuze van drie onafhankelijke lussen is:
</p>

<ul>
<li><b>Lus 1:</b> 1 – 2 – 3 – 4 – 1</li>
<li><b>Lus 2:</b> 3 – 5 – 7 – 4 – 3</li>
<li><b>Lus 3:</b> 5 – 6 – 8 – 7 – 5</li>
</ul>

<p>
Deze lussen zijn topologisch onafhankelijk:
geen van deze lussen kan geschreven worden als combinatie van de andere twee,
en samen leggen ze alle interne geometrische beperkingen van het mechanisme op.
</p>

<h4>Geometrische beschrijving op niveau van gewrichtspunten</h4>

<p>
Uitgedrukt in termen van de gewrichtspunten worden deze drie lussen:
</p>

<ul>
<li><b>Lus 1:</b> C – A/B – D – E – C</li>
<li><b>Lus 2:</b> E – D – F - G – H – E</li>
<li><b>Lus 3:</b> G – F – I – J – H – G</li>
</ul>

<p>
Hierbij wordt <b>A/B</b> gebruikt om aan te geven dat de punten <b>A</b> en <b>B</b>
samenvallen: topologisch horen ze bij twee verschillende gewrichten
(prismatic tussen 1–2 in A en revolute tussen 2–3 in B),
maar geometrisch blijven ze steeds hetzelfde punt in het vlak.
Er is dus <b>geen afzonderlijk lengtesegment AB</b>.
</p>

<h4>Interpretatie van de gekozen lussen</h4>

<ul>
<li>
In <b>lus 1</b> wordt de verbinding tussen de vaste stok, de schuiver,
link 3 en link 4 beschreven.  
Deze lus koppelt de invoercoördinaat van de schuiver rechtstreeks aan de hoeken
van de bovenste structuur.
</li>

<li>
In <b>lus 2</b> wordt de middelste schaarlus beschreven, gevormd door links
3, 5, 7 en 4.  
Hierin komen de interne koppelingen tussen de bovenste en onderste structuur tot uiting.
</li>

<li>
In <b>lus 3</b> wordt de buitenste lus beschreven, gevormd door links
5, 6, 8 en 7.  
Deze lus bepaalt de koppeling naar de buitenste rib van de paraplu.
</li>

<li>
Punt <b>K</b> wordt niet opgenomen in de sluitingslussen,
omdat het geen gewricht is maar een bijkomend punt op link 8
dat als <b>outputpunt</b> dient.
</li>
</ul>

<h4>Waarom deze drie lussen voldoende zijn</h4>

<p>
Aangezien het mechanisme drie onafhankelijke lussen bevat,
leveren deze drie lussen samen precies de nodige geometrische sluitingsvoorwaarden.
Elke lus geeft één vectoriële sluitingsvergelijking,
dus twee scalaire vergelijkingen na projectie op de x- en y-as.
</p>

<p>
In totaal ontstaan zo:
</p>

<ul>
<li><b>3 vectorvergelijkingen</b>, of equivalent</li>
<li><b>6 scalaire vergelijkingen</b>.</li>
</ul>

<p>
Deze zes vergelijkingen zijn precies nodig om de zes onbekende oriëntatiehoeken
van de bewegende links te bepalen:
<b>θ<sub>3</sub>, θ<sub>4</sub>, θ<sub>5</sub>, θ<sub>6</sub>, θ<sub>7</sub>, θ<sub>8</sub></b>,
voor een gegeven invoerpositie van de schuiver.
</p>

<h4>Conclusie</h4>

De kinematische analyse van dit paraplu-mechanisme wordt dus opgebouwd
op basis van <span style="color: Maroon;"><b>drie onafhankelijke kinematische lussen</b></span>:

Uitgedrukt op het niveau van de gewrichtspunten zijn dit respectievelijk:

Deze keuze is minimaal, volledig en consistent met de topologie van het mechanisme.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

%matplotlib widget

# ==========================================================
# Parameters van het paraplu-mechanisme
# Deze cel bevat enkel gegevens die tijdens het examen
# snel aangepast kunnen worden.
# ==========================================================

# Mastlengte
L1 = 2.5  # [m]

# Geometrische parameters van de links [m]
r3a = 0.760956
r3b = 0.175872
r4a = 0.955804
r4b = 0.590527
r5a = 0.283160
r5b = 0.307367
r6  = 0.662728
r7a = 0.175872
r7b = 0.486856
r8a = 0.307367
r8b = 0.662508

# Simulatie-instellingen
t_begin = 0.0     # begintijd [s]
t_end   = 12.0    # eindtijd [s]
Ts      = 0.05    # tijdstap [s]
t = np.arange(t_begin, t_end + Ts, Ts)

# Invoercoördinaat s:
# s wordt naar beneden gemeten vanaf punt C
# grote s = meer gesloten toestand
# kleine s = meer open toestand
s_lower = 0.125000    # minimale schuiverpositie [m]
s_upper = 1.875000    # maximale schuiverpositie [m]

# Tijdswet voor de schuiverbeweging
omega = 0.8       # pulsatie van de invoerwet [rad/s]
s_mid = 0.5 * (s_lower + s_upper)
A_s   = 0.5 * (s_upper - s_lower)

s   = s_mid + A_s * np.cos(omega * t)
ds  = -omega * A_s * np.sin(omega * t)
dds = -omega**2 * A_s * np.cos(omega * t)

# Initiële gok voor de positie-oplosser
# Deze beginwaarden bepalen mee welke configuratietak gevolgd wordt
theta3_init = 1.432805
theta4_init = -1.435562
theta5_init = -1.211154
theta6_init = -4.079521
theta7_init = -4.079521
theta8_init = -1.211154
```

<h3>Positie-analyse gebaseerd op sluitingsvergelijkingen</h3>

<p>
<span style="color: Maroon;"><b>Sluitingsvergelijkingen</b></span> drukken uit dat de links in het mechanisme
tijdens de beweging steeds met elkaar verbonden blijven.
Voor elke <b>onafhankelijke gesloten lus</b> geldt daarom dat de som van de
verplaatsingsvectoren over die lus gelijk is aan nul.
</p>

<p>
In dit mechanisme is de invoer de schuiverpositie <b>s</b>, en de onbekenden in de
positie-analyse zijn de zes oriëntatiehoeken
<b>θ<sub>3</sub>, θ<sub>4</sub>, θ<sub>5</sub>, θ<sub>6</sub>, θ<sub>7</sub></b> en <b>θ<sub>8</sub></b>.
Daarom zijn <b>zes scalaire vergelijkingen</b> nodig.
Die worden bekomen uit <b>drie onafhankelijke kinematische lussen</b>,
waarvan elke lus één vectoriële of twee scalaire vergelijkingen oplevert.
</p>

<p>
Zoals in les 2 worden alle oriëntatiehoeken gemeten ten opzichte van de globale x-as,
positief in tegenwijzerzin.
</p>

<p>
Belangrijk:
delen met hetzelfde linknummer, zoals <b>3a</b> en <b>3b</b>, of <b>8a</b> en <b>8b</b>,
behoren tot <b>éénzelfde star lichaam</b>.
Het zijn dus geen afzonderlijke links, maar vaste deelvectoren binnen dezelfde starre staaf.
</p>

<p>
We gebruiken de volgende drie onafhankelijke lussen:
</p>

<ul>
<li><b>lus 1:</b> C – A/B – D – E – C</li>
<li><b>lus 2:</b> E – D – F – G – H – E</li>
<li><b>lus 3:</b> G – F – I – J – H – G</li>
</ul>

<p>
Hierbij wordt <b>A/B</b> gebruikt om aan te geven dat de punten <b>A</b> en <b>B</b>
geometrisch samenvallen.
Topologisch behoren ze tot twee verschillende gewrichten
(prismatic tussen links 1–2 in A en revolute tussen links 2–3 in B),
maar in de vlakke geometrie is er <b>geen afzonderlijk segment AB</b>.
Dus geldt:
</p>

<p style="text-align:center;">
<b>\(\mathbf{r}_{AB} = \mathbf{0}\)</b>
</p>

<h4>Lus 1</h4>

<p>
Voor de eerste lus schrijven we:
</p>

$$
\mathbf{r}_{CA/B} + \mathbf{r}_{A/BD} + \mathbf{r}_{DE} + \mathbf{r}_{EC} = 0
$$

<p>
Omdat A en B samenvallen, kan deze vergelijking ook geschreven worden als:
</p>

$$
\mathbf{r}_{CA} + \mathbf{r}_{BD} + \mathbf{r}_{DE} + \mathbf{r}_{EC} = 0
$$

<p>
Met:
</p>

<ul>
<li>\(\mathbf{r}_{CA} = \begin{bmatrix}0 \\ -s \end{bmatrix}\), als de schuiverpositie \(s\) naar beneden
wordt gemeten vanaf punt \(C\);</li>
<li>\(\mathbf{r}_{BD} = r_{3a}\begin{bmatrix}\cos\theta_3 \\ \sin\theta_3 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{DE} = r_{3b}\begin{bmatrix}\cos\theta_3 \\ \sin\theta_3 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{EC} = -\,r_{4a}\begin{bmatrix}\cos\theta_4 \\ \sin\theta_4 \end{bmatrix}\).</li>
</ul>

<p>
Projectie op de globale x- en y-as geeft:
</p>

$$
\begin{aligned}
X &: r_{3a}\cos(\theta_3) + r_{3b}\cos(\theta_3) - r_{4a}\cos(\theta_4) = 0,
\\[4pt]
Y &: -s + r_{3a}\sin(\theta_3) + r_{3b}\sin(\theta_3) - r_{4a}\sin(\theta_4) = 0.
\end{aligned}
$$

<p>
Dus:
</p>

$$
\begin{aligned}
X &: (r_{3a}+r_{3b})\cos(\theta_3) - r_{4a}\cos(\theta_4) = 0,
\\[4pt]
Y &: -s + (r_{3a}+r_{3b})\sin(\theta_3) - r_{4a}\sin(\theta_4) = 0.
\end{aligned}
$$

<h4>Lus 2</h4>

<p>
Voor de tweede lus schrijven we:
</p>

$$
\mathbf{r}_{ED} + \mathbf{r}_{DF} + \mathbf{r}_{FG} + \mathbf{r}_{GH} + \mathbf{r}_{HE} = 0
$$

<p>
De overeenkomstige vectoren zijn:
</p>

<ul>
<li>\(\mathbf{r}_{ED} = -\,r_{3b}\begin{bmatrix}\cos\theta_3 \\ \sin\theta_3 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{DF} = r_{5a}\begin{bmatrix}\cos\theta_5 \\ \sin\theta_5 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{FG} = r_{5b}\begin{bmatrix}\cos\theta_5 \\ \sin\theta_5 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{GH} = r_{7a}\begin{bmatrix}\cos\theta_7 \\ \sin\theta_7 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{HE} = -\,r_{4b}\begin{bmatrix}\cos\theta_4 \\ \sin\theta_4 \end{bmatrix}\).</li>
</ul>

<p>
Projectie op de globale x- en y-as geeft:
</p>

$$
\begin{aligned}
X &: -r_{3b}\cos(\theta_3) + r_{5a}\cos(\theta_5) + r_{5b}\cos(\theta_5)
     + r_{7a}\cos(\theta_7) - r_{4b}\cos(\theta_4) = 0,
\\[4pt]
Y &: -r_{3b}\sin(\theta_3) + r_{5a}\sin(\theta_5) + r_{5b}\sin(\theta_5)
     + r_{7a}\sin(\theta_7) - r_{4b}\sin(\theta_4) = 0.
\end{aligned}
$$

<p>
Of samengenomen:
</p>

$$
\begin{aligned}
X &: -r_{3b}\cos(\theta_3) + (r_{5a}+r_{5b})\cos(\theta_5)
     + r_{7a}\cos(\theta_7) - r_{4b}\cos(\theta_4) = 0,
\\[4pt]
Y &: -r_{3b}\sin(\theta_3) + (r_{5a}+r_{5b})\sin(\theta_5)
     + r_{7a}\sin(\theta_7) - r_{4b}\sin(\theta_4) = 0.
\end{aligned}
$$

<h4>Lus 3</h4>

<p>
Voor de derde lus schrijven we:
</p>

$$
\mathbf{r}_{GF} + \mathbf{r}_{FI} + \mathbf{r}_{IJ} + \mathbf{r}_{JH} + \mathbf{r}_{HG} = 0
$$

<p>
De overeenkomstige vectoren zijn:
</p>

<ul>
<li>\(\mathbf{r}_{GF} = -\,r_{5b}\begin{bmatrix}\cos\theta_5 \\ \sin\theta_5 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{FI} = r_{6}\begin{bmatrix}\cos\theta_6 \\ \sin\theta_6 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{IJ} = r_{8a}\begin{bmatrix}\cos\theta_8 \\ \sin\theta_8 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{JH} = -\,r_{7b}\begin{bmatrix}\cos\theta_7 \\ \sin\theta_7 \end{bmatrix}\);</li>
<li>\(\mathbf{r}_{HG} = -\,r_{7a}\begin{bmatrix}\cos\theta_7 \\ \sin\theta_7 \end{bmatrix}\).</li>
</ul>

<p>
Projectie op de globale x- en y-as geeft:
</p>

$$
\begin{aligned}
X &: -r_{5b}\cos(\theta_5) + r_6\cos(\theta_6) + r_{8a}\cos(\theta_8)
     - r_{7b}\cos(\theta_7) - r_{7a}\cos(\theta_7) = 0,
\\[4pt]
Y &: -r_{5b}\sin(\theta_5) + r_6\sin(\theta_6) + r_{8a}\sin(\theta_8)
     - r_{7b}\sin(\theta_7) - r_{7a}\sin(\theta_7) = 0.
\end{aligned}
$$

<p>
Of samengenomen:
</p>

$$
\begin{aligned}
X &: -r_{5b}\cos(\theta_5) + r_6\cos(\theta_6) + r_{8a}\cos(\theta_8)
     - (r_{7a}+r_{7b})\cos(\theta_7) = 0,
\\[4pt]
Y &: -r_{5b}\sin(\theta_5) + r_6\sin(\theta_6) + r_{8a}\sin(\theta_8)
     - (r_{7a}+r_{7b})\sin(\theta_7) = 0.
\end{aligned}
$$

<h4>Samenvatting van het stelsel</h4>

<p>
De drie gekozen onafhankelijke lussen leveren samen een stelsel van
<b>zes niet-lineaire scalaire vergelijkingen</b> in de zes onbekende hoeken
<b>θ<sub>3</sub>, θ<sub>4</sub>, θ<sub>5</sub>, θ<sub>6</sub>, θ<sub>7</sub></b> en <b>θ<sub>8</sub></b>,
voor een gegeven schuiverpositie <b>s</b>.
</p>

<p>
Deze vergelijkingen vormen de basis van de numerieke positie-analyse.
Omdat het stelsel niet-lineair is, wordt het opgelost met een numerieke solver,
bijvoorbeeld <tt>fsolve</tt> uit <tt>scipy.optimize</tt>.
Daarbij vertrekt men van een geschikte beginwaarde voor de onbekende hoeken,
waarna de configuratie van het mechanisme voor elke opgelegde waarde van <b>s</b> berekend wordt.
</p>

<h4>Conclusie</h4>

<p>
De positie-analyse van het paraplu-mechanisme wordt dus opgebouwd uit
drie onafhankelijke sluitingslussen.
Door elke lus vectorieel te sluiten en vervolgens op de x- en y-as te projecteren,
ontstaat een volledig stelsel van vergelijkingen dat de configuratie van het mechanisme
voor een gegeven invoerpositie van de schuiver bepaalt.
</p>

</div>

```python
# Hulpfuncties en kinematische analyse van het paraplu-mechanisme

def rotate_vector(z, theta):
    rotation_matrix = np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])
    return np.dot(rotation_matrix, z)

def loop_closure_eqs(theta_init, s_k, r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a):
    theta3, theta4, theta5, theta6, theta7, theta8 = theta_init
    # Lus 1
    F1 = (r3a + r3b) * np.cos(theta3) - r4a * np.cos(theta4)
    F2 = -s_k + (r3a + r3b) * np.sin(theta3) - r4a * np.sin(theta4)
    # Lus 2
    F3 = -r3b * np.cos(theta3) + (r5a + r5b) * np.cos(theta5) + r7a * np.cos(theta7) - r4b * np.cos(theta4)
    F4 = -r3b * np.sin(theta3) + (r5a + r5b) * np.sin(theta5) + r7a * np.sin(theta7) - r4b * np.sin(theta4)
    # Lus 3
    F5 = -r5b * np.cos(theta5) + r6 * np.cos(theta6) + r8a * np.cos(theta8) - (r7a + r7b) * np.cos(theta7)
    F6 = -r5b * np.sin(theta5) + r6 * np.sin(theta6) + r8a * np.sin(theta8) - (r7a + r7b) * np.sin(theta7)
    return [F1, F2, F3, F4, F5, F6]

def get_jacobian(theta, r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a):
    theta3, theta4, theta5, theta6, theta7, theta8 = theta
    A = np.array([
        [-(r3a + r3b) * np.sin(theta3),  r4a * np.sin(theta4), 0, 0, 0, 0],
        [ (r3a + r3b) * np.cos(theta3), -r4a * np.cos(theta4), 0, 0, 0, 0],
        [ r3b * np.sin(theta3),          r4b * np.sin(theta4), -(r5a + r5b) * np.sin(theta5), 0, -r7a * np.sin(theta7), 0],
        [-r3b * np.cos(theta3),         -r4b * np.cos(theta4),  (r5a + r5b) * np.cos(theta5), 0,  r7a * np.cos(theta7), 0],
        [0, 0,  r5b * np.sin(theta5), -r6 * np.sin(theta6),  (r7a + r7b) * np.sin(theta7), -r8a * np.sin(theta8)],
        [0, 0, -r5b * np.cos(theta5),  r6 * np.cos(theta6), -(r7a + r7b) * np.cos(theta7),  r8a * np.cos(theta8)]
    ])
    return A

def get_B_vel(ds_k):
    return np.array([0, ds_k, 0, 0, 0, 0])

def get_B_acc(theta, dtheta, dds_k, r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a):
    theta3, theta4, theta5, theta6, theta7, theta8 = theta
    dt3, dt4, dt5, dt6, dt7, dt8 = dtheta
    B = np.array([
        (r3a + r3b) * np.cos(theta3) * dt3**2 - r4a * np.cos(theta4) * dt4**2,
        dds_k + (r3a + r3b) * np.sin(theta3) * dt3**2 - r4a * np.sin(theta4) * dt4**2,
        -r3b * np.cos(theta3) * dt3**2 + (r5a + r5b) * np.cos(theta5) * dt5**2 + r7a * np.cos(theta7) * dt7**2 - r4b * np.cos(theta4) * dt4**2,
        -r3b * np.sin(theta3) * dt3**2 + (r5a + r5b) * np.sin(theta5) * dt5**2 + r7a * np.sin(theta7) * dt7**2 - r4b * np.sin(theta4) * dt4**2,
        -r5b * np.cos(theta5) * dt5**2 + r6 * np.cos(theta6) * dt6**2 + r8a * np.cos(theta8) * dt8**2 - (r7a + r7b) * np.cos(theta7) * dt7**2,
        -r5b * np.sin(theta5) * dt5**2 + r6 * np.sin(theta6) * dt6**2 + r8a * np.sin(theta8) * dt8**2 - (r7a + r7b) * np.sin(theta7) * dt7**2
    ])
    return B

def kinematics_umbrella(r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a, s, ds, dds, theta3_init, theta4_init, theta5_init, theta6_init, theta7_init, theta8_init, t):
    optim_options = {"full_output": True}
    N = len(t)
    dt = t[1] - t[0] if N > 1 else 0.05
    
    # Gebruik opslagarrays
    th3, th4, th5, th6, th7, th8 = (np.zeros(N) for _ in range(6))
    dth3, dth4, dth5, dth6, dth7, dth8 = (np.zeros(N) for _ in range(6))
    ddth3, ddth4, ddth5, ddth6, ddth7, ddth8 = (np.zeros(N) for _ in range(6))
    cnd, res_pos = np.zeros(N), np.zeros(N)

    for k, s_k in enumerate(s):
        # Positie-analyse
        x, _, ier, message = fsolve(
            lambda x: loop_closure_eqs(x, s_k, r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a),
            [theta3_init, theta4_init, theta5_init, theta6_init, theta7_init, theta8_init],
            **optim_options
        )
        if ier != 1:
            print(f"Waarschuwing: fsolve convergeerde niet op stap {k}, s = {s_k:.6f} m")

        theta_k = x
        th3[k], th4[k], th5[k], th6[k], th7[k], th8[k] = theta_k
        res_pos[k] = np.linalg.norm(loop_closure_eqs(theta_k, s_k, r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a))

        # Matrices opbouwen
        J = get_jacobian(theta_k, r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a)
        cnd[k] = np.linalg.cond(J)

        # Snelheidsanalyse
        Bv = get_B_vel(ds[k])
        dq = np.linalg.solve(J, Bv)
        dth3[k], dth4[k], dth5[k], dth6[k], dth7[k], dth8[k] = dq

        # Versnellingsanalyse
        Ba = get_B_acc(theta_k, dq, dds[k], r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a)
        ddq = np.linalg.solve(J, Ba)
        ddth3[k], ddth4[k], ddth5[k], ddth6[k], ddth7[k], ddth8[k] = ddq

        # Branch tracking via vorige oplossing
        theta3_init, theta4_init, theta5_init, theta6_init, theta7_init, theta8_init = theta_k + dt * dq
        
    return th3, th4, th5, th6, th7, th8,            dth3, dth4, dth5, dth6, dth7, dth8,            ddth3, ddth4, ddth5, ddth6, ddth7, ddth8,            cnd, res_pos

# Uitvoeren van de volledige kinematische analyse
[theta3, theta4, theta5, theta6, theta7, theta8,
 dtheta3, dtheta4, dtheta5, dtheta6, dtheta7, dtheta8,
 ddtheta3, ddtheta4, ddtheta5, ddtheta6, ddtheta7, ddtheta8,
 cond, residual_pos] = kinematics_umbrella(
    r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a,
    s, ds, dds,
    theta3_init, theta4_init, theta5_init, theta6_init, theta7_init, theta8_init,
    t
)

# -------------------------------------------------------------
# CARTESISCHE COÖRDINATEN BEPALEN & VERIFICATIE VAN LUSSLUITING
# -------------------------------------------------------------
pt_names = ['C', 'B', 'D', 'E', 'F', 'H', 'G', 'I', 'J', 'K']
pts = {name: np.zeros((len(t), 2)) for name in pt_names}

for k in range(len(t)):
    C = np.array([0.0, 0.0])
    B = C + np.array([0.0, -s[k]])

    # Link 3
    D_from_3 = B + rotate_vector(np.array([r3a, 0.0]), theta3[k])
    E_from_3 = D_from_3 + rotate_vector(np.array([r3b, 0.0]), theta3[k])

    # Link 4
    E_from_4 = C + rotate_vector(np.array([r4a, 0.0]), theta4[k])
    H = E_from_4 + rotate_vector(np.array([r4b, 0.0]), theta4[k])
    
    # Link 5
    F = D_from_3 + rotate_vector(np.array([r5a, 0.0]), theta5[k])
    G_from_5 = F + rotate_vector(np.array([r5b, 0.0]), theta5[k])
    
    # Link 6
    I = F + rotate_vector(np.array([r6, 0.0]), theta6[k])

    # Link 7
    J_from_7 = H + rotate_vector(np.array([r7b, 0.0]), theta7[k])
    G_from_7 = H - rotate_vector(np.array([r7a, 0.0]), theta7[k])

    # Link 8
    J_from_8 = I + rotate_vector(np.array([r8a, 0.0]), theta8[k])
    K = J_from_7 + rotate_vector(np.array([r8b, 0.0]), theta8[k])
    
    pts['C'][k], pts['B'][k], pts['H'][k] = C, B, H
    pts['F'][k], pts['I'][k], pts['K'][k] = F, I, K
    # Gemiddelde punten om plotten soepel te maken
    pts['D'][k] = D_from_3
    pts['E'][k] = 0.5 * (E_from_3 + E_from_4)
    pts['G'][k] = 0.5 * (G_from_5 + G_from_7)
    pts['J'][k] = 0.5 * (J_from_7 + J_from_8)

# Lussluiting expliciet controleren (Cartesisch)
err_E = np.max(np.linalg.norm(pts['E'] - (pts['B'] + np.column_stack([np.cos(theta3)*(r3a+r3b), np.sin(theta3)*(r3a+r3b)])), axis=1))
err_G = np.max(np.linalg.norm(pts['G'] - (pts['H'] - np.column_stack([np.cos(theta7)*r7a, np.sin(theta7)*r7a])), axis=1))
err_J = np.max(np.linalg.norm(pts['J'] - (pts['I'] + np.column_stack([np.cos(theta8)*r8a, np.sin(theta8)*r8a])), axis=1))

print(f"Maximale Cartesische lussluitingsfouten:")
print(f"  Punt E (Link 3 vs Link 4): {err_E:.2e} m")
print(f"  Punt G (Link 5 vs Link 7): {err_G:.2e} m")
print(f"  Punt J (Link 7 vs Link 8): {err_J:.2e} m")
```

<h3>Controle van de positie-oplossing</h3>

<p>
Na het numeriek oplossen van de sluitingsvergelijkingen is het belangrijk om na te gaan
of de bekomen positie-oplossing ook <b>fysisch consistent</b> is over het volledige bewegingsbereik.
Een numerieke solver zoals <tt>fsolve</tt> levert immers voor elke opgelegde waarde van de invoercoördinaat
een oplossing voor de onbekende hoeken, maar die oplossing moet ook overeenkomen met
een realiseerbare beweging van het mechanisme.
</p>

<p>
Bij de positie-analyse zijn twee aandachtspunten belangrijk.
Ten eerste moeten de sluitingsvergelijkingen effectief goed voldaan zijn,
wat betekent dat de resterende fout op de lusvergelijkingen klein moet blijven.
Ten tweede moet de berekende configuratie op eenzelfde <b>oplossingsbranch</b> blijven.
Bij mechanismen met meerdere gesloten lussen kunnen immers verschillende configuraties bestaan
voor eenzelfde waarde van de invoer.
</p>

<p>
Daarom wordt in deze notebook telkens de oplossing van de vorige stap gebruikt als beginwaarde
voor de volgende stap.
Op die manier volgt de numerieke oplossing één continue configuratietak,
wat overeenkomt met een fysisch doorlopende open- en sluitbeweging van de paraplu.
</p>

<p>
In de volgende pythoncel wordt de positie-oplossing gevisualiseerd.
Aan de hand van de evolutie van de hoeken en van de outputpositie van punt <b>K</b>
kan gecontroleerd worden of:
</p>

<ul>
<li>de oplossing glad en continu varieert met de invoer;</li>
<li>er geen plotse sprongen tussen verschillende configuratietakken optreden;</li>
<li>de beweging van het mechanisme fysisch logisch overeenkomt met openen en sluiten.</li>
</ul>

<p>
De visualisatie van de positie-oplossing vormt dus een eerste kwalitatieve controle
vóór de animatie en vóór de verdere snelheids- en versnellingsanalyse.
</p>

</div>

```python
# Bereken de volledige kinematica van het mechanisme

t_size = len(t)
sim_fraction = 2
frames = t_size / sim_fraction
delta = int(np.floor(t_size / frames))
index_vec = np.arange(0, t_size, delta)

theta3   = np.zeros_like(t)
theta4   = np.zeros_like(t)
theta5   = np.zeros_like(t)
theta6   = np.zeros_like(t)
theta7   = np.zeros_like(t)
theta8   = np.zeros_like(t)

dtheta3  = np.zeros_like(t)
dtheta4  = np.zeros_like(t)
dtheta5  = np.zeros_like(t)
dtheta6  = np.zeros_like(t)
dtheta7  = np.zeros_like(t)
dtheta8  = np.zeros_like(t)

ddtheta3 = np.zeros_like(t)
ddtheta4 = np.zeros_like(t)
ddtheta5 = np.zeros_like(t)
ddtheta6 = np.zeros_like(t)
ddtheta7 = np.zeros_like(t)
ddtheta8 = np.zeros_like(t)

cond = np.zeros_like(t)
residual_pos = np.zeros_like(t)

[theta3, theta4, theta5, theta6, theta7, theta8,
 dtheta3, dtheta4, dtheta5, dtheta6, dtheta7, dtheta8,
 ddtheta3, ddtheta4, ddtheta5, ddtheta6, ddtheta7, ddtheta8,
 cond, residual_pos] = kinematics_umbrella(
    r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a,
    s, ds, dds,
    theta3_init, theta4_init, theta5_init, theta6_init, theta7_init, theta8_init,
    t
)

# Bepaal de outputpositie van punt K
Kx = np.zeros_like(t)
Ky = np.zeros_like(t)

for k in range(len(t)):
    C = np.array([0.0, 0.0])
    B = C + np.array([0.0, -s[k]])
    D = B + rotate_vector(np.array([r3a, 0.0]), theta3[k])
    F = D + rotate_vector(np.array([r5a, 0.0]), theta5[k])
    I = F + rotate_vector(np.array([r6, 0.0]), theta6[k])
    J = I + rotate_vector(np.array([r8a, 0.0]), theta8[k])
    K = J + rotate_vector(np.array([r8b, 0.0]), theta8[k])

    Kx[k] = K[0]
    Ky[k] = K[1]

# Positieplots
fig1, ax1 = plt.subplots(nrows=2, ncols=2, constrained_layout=True, figsize=(10, 8))
fig1.suptitle("Controle van de positie-oplossing")

ax1[0,0].plot(s, theta3, label=r'$\theta_3$')
ax1[0,0].plot(s, theta4, label=r'$\theta_4$')
ax1[0,0].plot(s, theta5, label=r'$\theta_5$')
ax1[0,0].plot(s, theta6, label=r'$\theta_6$')
ax1[0,0].plot(s, theta7, label=r'$\theta_7$')
ax1[0,0].plot(s, theta8, label=r'$\theta_8$')
ax1[0,0].set_xlabel(r'$s$ [m]')
ax1[0,0].set_ylabel(r'$\theta_i$ [rad]')
ax1[0,0].legend()

ax1[0,1].plot(s, Kx, label=r'$K_x$')
ax1[0,1].plot(s, Ky, label=r'$K_y$')
ax1[0,1].set_xlabel(r'$s$ [m]')
ax1[0,1].set_ylabel(r'$K_x, K_y$ [m]')
ax1[0,1].legend()

ax1[1,0].plot(Kx, Ky)
ax1[1,0].set_xlabel(r'$K_x$ [m]')
ax1[1,0].set_ylabel(r'$K_y$ [m]')
ax1[1,0].set_title("Traject van punt K")
ax1[1,0].axis('equal')

ax1[1,1].plot(s, residual_pos)
ax1[1,1].set_xlabel(r'$s$ [m]')
ax1[1,1].set_ylabel("closure residual [-]")
ax1[1,1].set_title("Numerieke sluitingsfout")

plt.show()

print(f"Maximale closure residual: {np.max(residual_pos):.3e}")
print(f"Gemiddelde closure residual: {np.mean(residual_pos):.3e}")
```

<h3>Snelheidsanalyse gebaseerd op sluitingsvergelijkingen</h3>

<p>
De snelheidsanalyse wordt bekomen door de <em>positie</em>-sluitingsvergelijkingen
naar de tijd af te leiden.
Omdat de geometrische lengtes van alle links constant zijn, zijn enkel de
oriëntatiehoeken van de links en de invoercoördinaat van de schuiver tijdsafhankelijk.
</p>

<p>
De invoer van het mechanisme is de schuiverpositie <b>s</b>, zodat de gekende invoersnelheid
gegeven wordt door <b>\(\dot{s}\)</b>.
De onbekenden in de snelheidsanalyse zijn de hoeksnelheden
<b>\(\omega_3, \omega_4, \omega_5, \omega_6, \omega_7, \omega_8\)</b>,
waarbij:
</p>

<p style="text-align:center;">
<b>\(\omega_i = \dot{\theta}_i\)</b>
</p>

<p>
De positievergelijkingen waarvan vertrokken wordt, zijn:
</p>

$$
\begin{aligned}
&\text{(lus 1)} \\[4pt]
X:\;& (r_{3a}+r_{3b})\cos(\theta_3) - r_{4a}\cos(\theta_4) = 0,
\\
Y:\;& -s + (r_{3a}+r_{3b})\sin(\theta_3) - r_{4a}\sin(\theta_4) = 0,
\\[10pt]

&\text{(lus 2)} \\[4pt]
X:\;& -r_{3b}\cos(\theta_3) + (r_{5a}+r_{5b})\cos(\theta_5)
      + r_{7a}\cos(\theta_7) - r_{4b}\cos(\theta_4) = 0,
\\
Y:\;& -r_{3b}\sin(\theta_3) + (r_{5a}+r_{5b})\sin(\theta_5)
      + r_{7a}\sin(\theta_7) - r_{4b}\sin(\theta_4) = 0,
\\[10pt]

&\text{(lus 3)} \\[4pt]
X:\;& -r_{5b}\cos(\theta_5) + r_6\cos(\theta_6) + r_{8a}\cos(\theta_8)
      - (r_{7a}+r_{7b})\cos(\theta_7) = 0,
\\
Y:\;& -r_{5b}\sin(\theta_5) + r_6\sin(\theta_6) + r_{8a}\sin(\theta_8)
      - (r_{7a}+r_{7b})\sin(\theta_7) = 0.
\end{aligned}
$$

<p>
Afleiden naar de tijd geeft de volgende snelheidsvergelijkingen.
</p>

<h4>Lus 1</h4>

$$
\begin{aligned}
X:\;& -(r_{3a}+r_{3b})\sin(\theta_3)\,\omega_3
      + r_{4a}\sin(\theta_4)\,\omega_4 = 0,
\\[6pt]
Y:\;& -\dot{s}
      + (r_{3a}+r_{3b})\cos(\theta_3)\,\omega_3
      - r_{4a}\cos(\theta_4)\,\omega_4 = 0.
\end{aligned}
$$

<h4>Lus 2</h4>

$$
\begin{aligned}
X:\;& r_{3b}\sin(\theta_3)\,\omega_3
      - (r_{5a}+r_{5b})\sin(\theta_5)\,\omega_5
      - r_{7a}\sin(\theta_7)\,\omega_7
      + r_{4b}\sin(\theta_4)\,\omega_4 = 0,
\\[6pt]
Y:\;& -r_{3b}\cos(\theta_3)\,\omega_3
      + (r_{5a}+r_{5b})\cos(\theta_5)\,\omega_5
      + r_{7a}\cos(\theta_7)\,\omega_7
      - r_{4b}\cos(\theta_4)\,\omega_4 = 0.
\end{aligned}
$$

<h4>Lus 3</h4>

$$
\begin{aligned}
X:\;& r_{5b}\sin(\theta_5)\,\omega_5
      - r_6\sin(\theta_6)\,\omega_6
      - r_{8a}\sin(\theta_8)\,\omega_8
      + (r_{7a}+r_{7b})\sin(\theta_7)\,\omega_7 = 0,
\\[6pt]
Y:\;& -r_{5b}\cos(\theta_5)\,\omega_5
      + r_6\cos(\theta_6)\,\omega_6
      + r_{8a}\cos(\theta_8)\,\omega_8
      - (r_{7a}+r_{7b})\cos(\theta_7)\,\omega_7 = 0.
\end{aligned}
$$

<h4>Structuur van het snelheidsstelsel</h4>

<p>
Dit levert een <b>lineair stelsel van zes vergelijkingen</b> in de zes onbekende
hoeksnelheden:
</p>

<p style="text-align:center;">
<b>\(\omega_3,\ \omega_4,\ \omega_5,\ \omega_6,\ \omega_7,\ \omega_8\)</b>
</p>

<p>
De invoer is de schuiversnelheid <b>\(\dot{s}\)</b>, die als gekend wordt beschouwd.
Voor een gegeven configuratie van het mechanisme, dus voor gekende waarden van
de hoeken \(\theta_3,\theta_4,\theta_5,\theta_6,\theta_7,\theta_8\),
kan dit lineaire stelsel numeriek opgelost worden om alle hoeksnelheden te bepalen.
</p>

<h4>Matrixvorm</h4>

<p>
Het snelheidsprobleem kan compact geschreven worden in matrixvorm als:
</p>

$$
A(\theta)\,\omega = B(\dot{s})
$$

<p>
waarbij:
</p>

<ul>
<li><b>A(\theta)</b> de configuratie-afhankelijke coëfficiëntenmatrix is;</li>
<li><b>\(\omega\)</b> de vector van onbekende hoeksnelheden is;</li>
<li><b>B(\dot{s})</b> het rechterlid is dat de opgelegde schuiversnelheid bevat.</li>
</ul>

<p>
Meer concreet is:
</p>

<p style="text-align:center;">
<b>\(\omega = [\omega_3,\ \omega_4,\ \omega_5,\ \omega_6,\ \omega_7,\ \omega_8]^T\)</b>
</p>

<p>
Deze formulering is bijzonder nuttig omdat ze rechtstreeks aansluit bij de numerieke implementatie:
voor elke gekende configuratie worden matrix <b>A</b> en vector <b>B</b> opgebouwd,
waarna de onbekende hoeksnelheden bepaald worden met een lineaire oplosser.
</p>

<h4>Fysische interpretatie en singulariteiten</h4>

<p>
De snelheidsanalyse is <b>configuratie-afhankelijk</b>:
de coëfficiënten van het stelsel veranderen voortdurend met de hoeken van het mechanisme.
Daardoor verandert ook de relatie tussen de opgelegde schuiversnelheid en de resulterende
hoeksnelheden van de links over het bewegingsbereik.
</p>

<p>
Wanneer de matrix <b>A</b> slecht geconditioneerd wordt, wijst dit op een configuratie
in de buurt van een <span style="color: Maroon;"><b>singulariteit of dood punt</b></span>.
In zulke configuraties kunnen kleine veranderingen in de invoer grote veranderingen in de
uitvoer veroorzaken, en worden de berekende snelheden numeriek gevoelig.
Daarom is de condition number van <b>A</b> een belangrijke indicator in de verdere analyse.
</p>

<h4>Conclusie</h4>

<p>
De snelheidsanalyse van het paraplu-mechanisme volgt rechtstreeks uit het afleiden
van de drie positie-sluitingsvergelijkingen.
Omdat de resulterende vergelijkingen lineair zijn in de onbekende hoeksnelheden,
kunnen deze efficiënt opgelost worden zodra de positie-analyse uitgevoerd is.
Zo wordt de schuiversnelheid \(\dot{s}\) gekoppeld aan de rotatiesnelheden van
alle bewegende links in het mechanisme.
</p>

</div>

```python
def compute_A_matrix(theta, r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a):
    theta3, theta4, theta5, theta6, theta7, theta8 = theta

    A = np.array([
        [-(r3a + r3b) * np.sin(theta3),  r4a * np.sin(theta4), 0, 0, 0, 0],
        [ (r3a + r3b) * np.cos(theta3), -r4a * np.cos(theta4), 0, 0, 0, 0],

        [ r3b * np.sin(theta3),          r4b * np.sin(theta4), -(r5a + r5b) * np.sin(theta5), 0, -r7a * np.sin(theta7), 0],
        [-r3b * np.cos(theta3),         -r4b * np.cos(theta4),  (r5a + r5b) * np.cos(theta5), 0,  r7a * np.cos(theta7), 0],

        [0, 0,  r5b * np.sin(theta5), -r6 * np.sin(theta6),  (r7a + r7b) * np.sin(theta7), -r8a * np.sin(theta8)],
        [0, 0, -r5b * np.cos(theta5),  r6 * np.cos(theta6), -(r7a + r7b) * np.cos(theta7),  r8a * np.cos(theta8)]
    ])

    return A
```

```python
# Snelheidsanalyse op basis van de reeds uitgevoerde kinematische analyse
# De hoeksnelheden dtheta_i zijn al berekend in kinematics_umbrella().
# In deze cel bouwen we expliciet A en B op, controleren we het snelheidsstelsel
# en bepalen we de relevante outputsnelheden van punt K.

def perp(v):
    """2D-operator voor omega x r = [-omega*y, omega*x]."""
    return np.array([-v[1], v[0]])

# opslag
vel_residual = np.zeros_like(t)
Kdot_x = np.zeros_like(t)
Kdot_y = np.zeros_like(t)
Kdot_norm = np.zeros_like(t)

for k in range(len(t)):
    theta_k = np.array([
        theta3[k],
        theta4[k],
        theta5[k],
        theta6[k],
        theta7[k],
        theta8[k]
    ])

    omega_k = np.array([
        dtheta3[k],
        dtheta4[k],
        dtheta5[k],
        dtheta6[k],
        dtheta7[k],
        dtheta8[k]
    ])

    # matrix A van het snelheidsstelsel
    A = compute_A_matrix(theta_k, r3a, r3b, r4a, r4b, r5a, r5b, r6, r7a, r7b, r8a)

    # rechterlid B
    B = np.array([
        0.0,
        ds[k],
        0.0,
        0.0,
        0.0,
        0.0
    ])

    # controle van het snelheidsstelsel A * omega = B
    vel_residual[k] = np.linalg.norm(A @ omega_k - B)

    # --- relevante puntsnelheden bepalen ---
    # Posities van de nodige punten
    C = np.array([0.0, 0.0])
    Bp = C + np.array([0.0, -s[k]])

    D = Bp + rotate_vector(np.array([r3a, 0.0]), theta3[k])
    F = D  + rotate_vector(np.array([r5a, 0.0]), theta5[k])
    I = F  + rotate_vector(np.array([r6, 0.0]),  theta6[k])
    J = I  + rotate_vector(np.array([r8a, 0.0]), theta8[k])
    K = J  + rotate_vector(np.array([r8b, 0.0]), theta8[k])

    # snelheden van opeenvolgende punten
    vB = np.array([0.0, -ds[k]])

    r_BD = D - Bp
    vD = vB + dtheta3[k] * perp(r_BD)

    r_DF = F - D
    vF = vD + dtheta5[k] * perp(r_DF)

    r_FI = I - F
    vI = vF + dtheta6[k] * perp(r_FI)

    r_IK = K - I
    vK = vI + dtheta8[k] * perp(r_IK)

    Kdot_x[k] = vK[0]
    Kdot_y[k] = vK[1]
    Kdot_norm[k] = np.linalg.norm(vK)

print("Maximale residual van het snelheidsstelsel:")
print(f"max ||A·omega - B|| = {np.max(vel_residual):.3e}")

print("\nMaximale norm van de snelheid van punt K:")
print(f"max |v_K| = {np.max(Kdot_norm):.6f} m/s")
```

<h3>Versnellingsanalyse gebaseerd op sluitingsvergelijkingen</h3>

<p>
De versnellingsanalyse wordt verkregen door de <em>snelheids</em>-sluitingsvergelijkingen
naar de tijd af te leiden.
Daarbij worden de hoeksnelheden \(\omega_i = \dot{\theta}_i\) opnieuw tijdsafhankelijk,
en ontstaan bijkomende termen in \(\omega_i^2\), afkomstig van de afgeleide van
sinus- en cosinustermen.
</p>

<p>
De onbekenden in deze analyse zijn de hoekversnellingen
<b>\(\alpha_3, \alpha_4, \alpha_5, \alpha_6, \alpha_7, \alpha_8\)</b>, met:
</p>

<p style="text-align:center;">
<b>\(\alpha_i = \dot{\omega}_i = \ddot{\theta}_i\)</b>
</p>

<p>
De invoer is de schuiverbeweging \(s(t)\), zodat naast de schuiversnelheid \(\dot{s}\)
nu ook de schuiverversnelling <b>\(\ddot{s}\)</b> als gekend wordt beschouwd.
</p>

<p>
We vertrekken van de snelheidsvergelijkingen:
</p>

$$
\begin{aligned}
&\text{(lus 1)} \\[4pt]
X:\;& -(r_{3a}+r_{3b})\sin(\theta_3)\,\omega_3
      + r_{4a}\sin(\theta_4)\,\omega_4 = 0,
\\
Y:\;& -\dot{s}
      + (r_{3a}+r_{3b})\cos(\theta_3)\,\omega_3
      - r_{4a}\cos(\theta_4)\,\omega_4 = 0,
\\[10pt]

&\text{(lus 2)} \\[4pt]
X:\;& r_{3b}\sin(\theta_3)\,\omega_3
      - (r_{5a}+r_{5b})\sin(\theta_5)\,\omega_5
      - r_{7a}\sin(\theta_7)\,\omega_7
      + r_{4b}\sin(\theta_4)\,\omega_4 = 0,
\\
Y:\;& -r_{3b}\cos(\theta_3)\,\omega_3
      + (r_{5a}+r_{5b})\cos(\theta_5)\,\omega_5
      + r_{7a}\cos(\theta_7)\,\omega_7
      - r_{4b}\cos(\theta_4)\,\omega_4 = 0,
\\[10pt]

&\text{(lus 3)} \\[4pt]
X:\;& r_{5b}\sin(\theta_5)\,\omega_5
      - r_6\sin(\theta_6)\,\omega_6
      - r_{8a}\sin(\theta_8)\,\omega_8
      + (r_{7a}+r_{7b})\sin(\theta_7)\,\omega_7 = 0,
\\
Y:\;& -r_{5b}\cos(\theta_5)\,\omega_5
      + r_6\cos(\theta_6)\,\omega_6
      + r_{8a}\cos(\theta_8)\,\omega_8
      - (r_{7a}+r_{7b})\cos(\theta_7)\,\omega_7 = 0.
\end{aligned}
$$

<p>
Afleiden naar de tijd geeft de versnellingsvergelijkingen.
</p>

<h4>Lus 1</h4>

$$
\begin{aligned}
0 =\;& -(r_{3a}+r_{3b})\cos(\theta_3)\,\omega_3^2
      -(r_{3a}+r_{3b})\sin(\theta_3)\,\alpha_3
\\
&\quad + r_{4a}\cos(\theta_4)\,\omega_4^2
      + r_{4a}\sin(\theta_4)\,\alpha_4,
\\[8pt]
0 =\;& -\ddot{s}
      -(r_{3a}+r_{3b})\sin(\theta_3)\,\omega_3^2
      +(r_{3a}+r_{3b})\cos(\theta_3)\,\alpha_3
\\
&\quad + r_{4a}\sin(\theta_4)\,\omega_4^2
      - r_{4a}\cos(\theta_4)\,\alpha_4.
\end{aligned}
$$

<h4>Lus 2</h4>

$$
\begin{aligned}
0 =\;& r_{3b}\cos(\theta_3)\,\omega_3^2
      + r_{3b}\sin(\theta_3)\,\alpha_3
\\
&\quad -(r_{5a}+r_{5b})\cos(\theta_5)\,\omega_5^2
      -(r_{5a}+r_{5b})\sin(\theta_5)\,\alpha_5
\\
&\quad -r_{7a}\cos(\theta_7)\,\omega_7^2
      - r_{7a}\sin(\theta_7)\,\alpha_7
\\
&\quad + r_{4b}\cos(\theta_4)\,\omega_4^2
      + r_{4b}\sin(\theta_4)\,\alpha_4,
\\[10pt]
0 =\;& r_{3b}\sin(\theta_3)\,\omega_3^2
      - r_{3b}\cos(\theta_3)\,\alpha_3
\\
&\quad -(r_{5a}+r_{5b})\sin(\theta_5)\,\omega_5^2
      +(r_{5a}+r_{5b})\cos(\theta_5)\,\alpha_5
\\
&\quad -r_{7a}\sin(\theta_7)\,\omega_7^2
      + r_{7a}\cos(\theta_7)\,\alpha_7
\\
&\quad + r_{4b}\sin(\theta_4)\,\omega_4^2
      - r_{4b}\cos(\theta_4)\,\alpha_4.
\end{aligned}
$$

<h4>Lus 3</h4>

$$
\begin{aligned}
0 =\;& r_{5b}\cos(\theta_5)\,\omega_5^2
      + r_{5b}\sin(\theta_5)\,\alpha_5
\\
&\quad - r_6\cos(\theta_6)\,\omega_6^2
      - r_6\sin(\theta_6)\,\alpha_6
\\
&\quad - r_{8a}\cos(\theta_8)\,\omega_8^2
      - r_{8a}\sin(\theta_8)\,\alpha_8
\\
&\quad + (r_{7a}+r_{7b})\cos(\theta_7)\,\omega_7^2
      + (r_{7a}+r_{7b})\sin(\theta_7)\,\alpha_7,
\\[10pt]
0 =\;& r_{5b}\sin(\theta_5)\,\omega_5^2
      - r_{5b}\cos(\theta_5)\,\alpha_5
\\
&\quad - r_6\sin(\theta_6)\,\omega_6^2
      + r_6\cos(\theta_6)\,\alpha_6
\\
&\quad - r_{8a}\sin(\theta_8)\,\omega_8^2
      + r_{8a}\cos(\theta_8)\,\alpha_8
\\
&\quad + (r_{7a}+r_{7b})\sin(\theta_7)\,\omega_7^2
      - (r_{7a}+r_{7b})\cos(\theta_7)\,\alpha_7.
\end{aligned}
$$

<h4>Structuur van het versnellingsstelsel</h4>

<p>
Dit resulteert opnieuw in een <b>lineair stelsel van zes vergelijkingen</b>
in de zes onbekende hoekversnellingen:
</p>

<p style="text-align:center;">
<b>\(\alpha_3,\ \alpha_4,\ \alpha_5,\ \alpha_6,\ \alpha_7,\ \alpha_8\)</b>
</p>

<p>
In matrixvorm kan dit geschreven worden als:
</p>

$$
A(\theta)\,\alpha = B(\theta,\omega,\ddot{s})
$$

<p>
Hierbij is <b>\(A(\theta)\)</b> dezelfde coëfficiëntenmatrix als in de snelheidsanalyse.
Het rechterlid <b>\(B\)</b> bevat de bekende termen in \(\omega_i^2\)
en de opgelegde invoerversnelling \(\ddot{s}\).
</p>

<h4>Interpretatie</h4>

<p>
De versnellingsanalyse beschrijft hoe de opgelegde translatieversnelling van de schuiver
wordt omgezet in hoekversnellingen van alle bewegende links.
Omdat de matrix \(A\) configuratie-afhankelijk is, hangen ook de berekende versnellingen
sterk af van de stand van het mechanisme.
</p>

<p>
Wanneer het mechanisme een configuratie nadert waarvoor de matrix \(A\)
slecht geconditioneerd of bijna singulier wordt, kunnen de berekende hoekversnellingen
zeer groot worden, zelfs bij beperkte invoerversnelling.
Daarom vormen grote versnellingspieken een belangrijke indicator voor kritische configuraties.
</p>

<h4>Conclusie</h4>

<p>
Door de snelheids-sluitingsvergelijkingen opnieuw naar de tijd af te leiden,
wordt een lineair stelsel bekomen voor de hoekversnellingen van alle bewegende links.
Samen met de positie- en snelheidsanalyse levert dit een volledige kinematische beschrijving
van het paraplu-mechanisme.
</p>

</div>

```python
# Versnellingsanalyse: afgeleide grootheden van punt K
# De hoekversnellingen ddtheta3 ... ddtheta8 zijn reeds berekend in kinematics_umbrella().

Kddot_x = np.zeros_like(t)
Kddot_y = np.zeros_like(t)
Kddot_norm = np.zeros_like(t)

for k in range(len(t)):
    theta8_k = theta8[k]
    dtheta8_k = dtheta8[k]
    ddtheta8_k = ddtheta8[k]

    # vectoren op link 8
    r_IJ = rotate_vector(np.array([r8a, 0.0]), theta8_k)
    r_JK = rotate_vector(np.array([r8b, 0.0]), theta8_k)
    r_IK = r_IJ + r_JK

    x_IK = r_IK[0]
    y_IK = r_IK[1]

    # punt I
    C = np.array([0.0, 0.0])
    B = C + np.array([0.0, -s[k]])
    D = B + rotate_vector(np.array([r3a, 0.0]), theta3[k])
    F = D + rotate_vector(np.array([r5a, 0.0]), theta5[k])
    I = F + rotate_vector(np.array([r6, 0.0]), theta6[k])

    # snelheid van I
    vx_B = 0.0
    vy_B = -ds[k]

    vx_D = vx_B - r3a * np.sin(theta3[k]) * dtheta3[k]
    vy_D = vy_B + r3a * np.cos(theta3[k]) * dtheta3[k]

    vx_F = vx_D - r5a * np.sin(theta5[k]) * dtheta5[k]
    vy_F = vy_D + r5a * np.cos(theta5[k]) * dtheta5[k]

    vx_I = vx_F - r6 * np.sin(theta6[k]) * dtheta6[k]
    vy_I = vy_F + r6 * np.cos(theta6[k]) * dtheta6[k]

    # versnelling van I
    ax_B = 0.0
    ay_B = -dds[k]

    ax_D = ax_B - r3a * (
        np.cos(theta3[k]) * dtheta3[k]**2 + np.sin(theta3[k]) * ddtheta3[k]
    )
    ay_D = ay_B + r3a * (
        -np.sin(theta3[k]) * dtheta3[k]**2 + np.cos(theta3[k]) * ddtheta3[k]
    )

    ax_F = ax_D - r5a * (
        np.cos(theta5[k]) * dtheta5[k]**2 + np.sin(theta5[k]) * ddtheta5[k]
    )
    ay_F = ay_D + r5a * (
        -np.sin(theta5[k]) * dtheta5[k]**2 + np.cos(theta5[k]) * ddtheta5[k]
    )

    ax_I = ax_F - r6 * (
        np.cos(theta6[k]) * dtheta6[k]**2 + np.sin(theta6[k]) * ddtheta6[k]
    )
    ay_I = ay_F + r6 * (
        -np.sin(theta6[k]) * dtheta6[k]**2 + np.cos(theta6[k]) * ddtheta6[k]
    )

    # versnelling van K = versnelling van I + relatieve versnelling door rotatie van link 8
    Kddot_x[k] = ax_I - ddtheta8_k * y_IK - dtheta8_k**2 * x_IK
    Kddot_y[k] = ay_I + ddtheta8_k * x_IK - dtheta8_k**2 * y_IK

    Kddot_norm[k] = np.sqrt(Kddot_x[k]**2 + Kddot_y[k]**2)
```

<h3>Analyse van singulariteiten en conditionering</h3>

<p>
Bij een vlak mechanisme verandert een <span style="color: Maroon;"><b>singulariteit</b></span>
in het algemeen de topologische mobiliteit niet:
het aantal vrijheidsgraden van het mechanisme blijft dus gelijk.
Toch hebben zulke configuraties een zeer grote invloed op het
<b>kinematisch gedrag</b> en op de <b>numerieke oplosbaarheid</b> van het probleem.
</p>

<p>
In de snelheidsanalyse wordt voor elke configuratie een lineair stelsel opgelost
van de vorm:
</p>

$$
A(\theta)\,x = B
$$

<p>
waarbij de matrix <b>A</b> afhangt van de actuele configuratie van het mechanisme.
Wanneer het mechanisme een singuliere of bijna-singuliere configuratie nadert,
wordt deze matrix slecht geconditioneerd.
Dat betekent dat kleine veranderingen in invoer of afrondingsfouten kunnen leiden
tot grote veranderingen in de berekende snelheden en versnellingen.
</p>

<p>
Een praktische indicator hiervoor is het
<span style="color: Maroon;"><b>condition number</b></span> van de matrix <b>A</b>.
</p>

<ul>
<li>een lage waarde van <b>cond(A)</b> wijst op een goed geconditioneerd stelsel;</li>
<li>een hoge waarde van <b>cond(A)</b> wijst op numerieke gevoeligheid;</li>
<li>zeer grote pieken kunnen wijzen op kritische configuraties of naderende singulariteiten.</li>
</ul>

<p>
Het is daarom nuttig om <b>cond(A)</b> over het volledige bewegingsbereik te onderzoeken.
Zo kan men nagaan:
</p>

<ul>
<li>in welke configuraties het mechanisme numeriek gevoelig wordt;</li>
<li>waar snelheden en versnellingen sterk kunnen toenemen;</li>
<li>of het gekozen bewegingsbereik veilig verwijderd blijft van kritische standen.</li>
</ul>

<p>

```python
# Analyse van conditionering en kritische configuraties

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. Basisstatistieken
# -----------------------------
cond_finite = cond[np.isfinite(cond)]

print("Statistiek van de condition number:")
print(f"minimum cond(A) = {np.min(cond_finite):.3e}")
print(f"gemiddelde cond(A) = {np.mean(cond_finite):.3e}")
print(f"maximum cond(A) = {np.max(cond_finite):.3e}")

# -----------------------------
# 2. Drempel voor kritische configuraties
# -----------------------------
# We kiezen een generieke drempel:
# ofwel absoluut groot,
# ofwel behorend tot de hoogste 5% van de cond-waarden.
cond_threshold_abs = 1e3
cond_threshold_percentile = np.percentile(cond_finite, 95)

cond_threshold = max(cond_threshold_abs, cond_threshold_percentile)

print(f"\nGebruikte drempel voor kritische configuraties: cond(A) > {cond_threshold:.3e}")

critical_idx = np.where(cond > cond_threshold)[0]

# -----------------------------
# 3. Kritische punten groeperen
# -----------------------------
# Opeenvolgende indices worden gegroepeerd tot één kritische zone.
critical_groups = []

if len(critical_idx) > 0:
    start = critical_idx[0]
    prev = critical_idx[0]

    for idx in critical_idx[1:]:
        if idx == prev + 1:
            prev = idx
        else:
            critical_groups.append((start, prev))
            start = idx
            prev = idx

    critical_groups.append((start, prev))

print(f"Aantal kritische zones: {len(critical_groups)}")

for i, (i_start, i_end) in enumerate(critical_groups, start=1):
    i_peak = i_start + np.argmax(cond[i_start:i_end+1])
    print(
        f"Zone {i}: "
        f"t ≈ {t[i_peak]:.3f} s, "
        f"s ≈ {s[i_peak]:.3f} m, "
        f"cond(A) ≈ {cond[i_peak]:.3e}"
    )

# -----------------------------
# 4. Plot cond(A) als functie van t
# -----------------------------
fig_cond_t, ax_cond_t = plt.subplots(figsize=(10, 4))
fig_cond_t.suptitle("Condition number van de snelheidsmatrix als functie van de tijd")

ax_cond_t.plot(t, cond, label='cond(A)')
ax_cond_t.axhline(cond_threshold, linestyle='--', label='kritische drempel')

for (i_start, i_end) in critical_groups:
    i_peak = i_start + np.argmax(cond[i_start:i_end+1])
    ax_cond_t.plot(t[i_peak], cond[i_peak], 'o')
    ax_cond_t.annotate(
        f"s={s[i_peak]:.3f} m",
        (t[i_peak], cond[i_peak]),
        textcoords="offset points",
        xytext=(5, 5),
        fontsize=9
    )

ax_cond_t.set_xlabel('t [s]')
ax_cond_t.set_ylabel('cond(A) [-]')
ax_cond_t.legend()
plt.show()

# -----------------------------
# 5. Plot cond(A) als functie van s
# -----------------------------
fig_cond_s, ax_cond_s = plt.subplots(figsize=(10, 4))
fig_cond_s.suptitle("Condition number van de snelheidsmatrix als functie van de schuiverpositie")

ax_cond_s.plot(s, cond, label='cond(A)')
ax_cond_s.axhline(cond_threshold, linestyle='--', label='kritische drempel')

for (i_start, i_end) in critical_groups:
    i_peak = i_start + np.argmax(cond[i_start:i_end+1])
    ax_cond_s.plot(s[i_peak], cond[i_peak], 'o')
    ax_cond_s.annotate(
        f"t={t[i_peak]:.2f} s",
        (s[i_peak], cond[i_peak]),
        textcoords="offset points",
        xytext=(5, 5),
        fontsize=9
    )

ax_cond_s.set_xlabel('s [m]')
ax_cond_s.set_ylabel('cond(A) [-]')
ax_cond_s.legend()
plt.show()
```

# 💡 Examenvoorbereiding: Theoretische Inzichten (Paraplu-Mechanisme)

De professor hecht veel belang aan het "hardop denken" als ingenieur. Hieronder staan de uitgewerkte antwoorden op de meest kritische theorievragen over dit specifieke mechanisme, zodat je hierop kan terugvallen tijdens het examen.

### 1. Wat als we de motor verplaatsen?
* **"Waarom zit de aandrijving op de centrale schuiver en niet bovenaan of in een scharnier?"**
  * Bij een paraplu is de centrale schuiver de enige plaats waar de mens (of een compacte lineaire motor) ergonomisch kracht kan uitoefenen. 
  * Als enkel *mechanische efficiëntie* telde, zouden we de aandrijving plaatsen op een stang die continu loodrecht (90° transmissiehoek) aangedreven kan worden, en waarvan de motor-massatraagheid zelf niet meebeweegt (bijv. een roterende motor op het vaste grondscharnier `C`). Door de motor op de translerende schuiver te zetten, beweegt de massa van de aandrijving mee, wat extra (onbalans)krachten genereert.

### 2. Dode Punten en Singulariteiten
* **"Heeft de paraplu een dood punt als hij volledig is uitgeklapt? Is dat negatief?"**
  * Ja, er treedt een dood punt (singulariteit) op als de stangen in elkaars verlengde liggen ($\mu \to 0^\circ$ of $180^\circ$). 
  * Dit is in dit geval **positief**! Dit heet een *over-center lock*. Hierdoor neemt het mechanische gestel zélf de externe krachten (zoals de wind die op de paraplu blaast) op in zijn langsrichting. De aandrijfmotor of de mens moet daardoor nul houdkracht of koppel leveren om de paraplu open te houden. (Vergelijkbaar met een landingsgestel van een vliegtuig).

### 3. Transmissiehoek vs. Conditiegetal (Condition Number)
* **"Waar is de overdracht van je krachten het best en het slechtst?"**
  * De overdracht is het best rond het midden van de beweging, waar de hoek $\mu$ tussen de stangen dichtbij $90^\circ$ ligt. (Zie de grafiek van de transmissiehoek hierboven).
  * Het is het slechtst aan de uitersten (volledig open of gesloten). Op dat moment zie je het **Condition Number** in de eerdere grafiek exponentieel pieken. Een hoog conditiegetal betekent dat de Jacobiaan bijna singulier is: kleine veranderingen in motorsnelheid leiden tot enorme (theoretisch oneindige) snelheden of krachten in de stangen.

### 4. Waarom kloppen sommige berekende krachten niet met de realiteit?
* **"Waarom zijn je berekende krachten in een ideaal model soms verdacht laag?"**
  * Omdat we in deze zuivere kinematische/virtuele-arbeid analyse **wrijving** en **zwaartekracht** (eigen massa van de stangen) verwaarlozen. In de realiteit zal de wrijving in de vele scharnieren (zeker als de drukhoek stijgt) de vereiste aandrijfkracht substantieel verhogen.
